```python
#Step1: read excel sheet
#Step 2: define groups and their corresponding alts
#Step 3: run loop with random implementation of alts
#Step 4: export new sheet

from openpyxl import load_workbook
import random
```

```python
wb0 = load_workbook(r"C:\Users\User\Desktop\Controle Clientes.xlsx")
ws0 = wb0.active
wb1 = load_workbook(r"C:\Users\User\Desktop\Atualização Controle Clientes-alt2.xlsx")
ws1 = wb1.active

vow1 = ["a", "e", "i"]
vow2 = ["o", "u"]
con1 = ["b", "c", "d", "f", "j", "p", "g", "k", "s", "t"]
con2 = ["h", "l", "m", "n", "v", "w", "x", "y", "z"]
vow3 = ["A", "E", "I"]
vow4 = ["O", "U"]
con3 = ["B", "C", "D", "F", "J", "P", "G", "K", "S", "T"]
con4 = ["H", "L", "M", "N", "V", "W", "X", "Y", "Z"]

def transform(text):
    result = ""
    for letter in text:
        if letter in vow1:
            result += random.choice(vow1)
        elif letter in vow2:
            result += random.choice(vow2)
        elif letter in vow3:
            result += random.choice(vow3)
        elif letter in vow4:
            result += random.choice(vow4)
        elif letter in con1:
            result += random.choice(con1)
        elif letter in con2:
            result += random.choice(con2)
        elif letter in con3:
            result += random.choice(con3)
        elif letter in con4:
            result += random.choice(con4)
        elif letter.isnumeric():
            result += random.choice("0123456789")
        else:
            result += letter
    return result

for row in ws0.iter_rows(min_row=2):
    for cell in row:
        if isinstance(cell.value, str):
            cell.value = transform(cell.value)
        elif isinstance(cell.value, (int, float)):
            cell.value = transform(str(int(cell.value)))
for row in ws1.iter_rows(min_row=2):
    for cell in row:
        if isinstance(cell.value, str):
            cell.value = transform(cell.value)
        elif isinstance(cell.value, (int, float)):
            cell.value = transform(str(int(cell.value)))


wb0.save(r"C:\Users\User\Documents\GitHub\CRMB2B\Gibb Original.xlsx")
wb1.save(r"C:\Users\User\Documents\GitHub\CRMB2B\Gibb Corrected.xlsx")
```
